```python
import pandas as pd
import seaborn as sns
```

### 1. Load Data & Data Quality Report

```python
# Load dataset
file = pd.read_csv('taitanic dataset.csv')
df = file.copy()
df
```

```python
# Data quality report
print('Duplicates:', df.duplicated().sum())
```

```python
print('\nNulls per column:\n', df.isnull().sum())
```

```python
print('\nData types:\n', df.dtypes)
```

```python
print('\nMin/Max anomalies:\n', df.describe().loc[['min', 'max']])
```

### 2. Missing Data Handling

* deck: Dropped because over 75% of the data is missing.
* age: Filled with the median because age has outliers.
* embarked: Filled with the mode (most common port) since only 2 rows are missing.

```python
df = df.drop(columns=['deck'], errors='ignore')
df
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Age']
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df['Embarked']
```

### 3. Duplicate Removal

```python
before = len(df)
df = df.drop_duplicates()
print('Duplicates removed:', before - len(df))
```

### 4. Standardisation

```python
# Fix capitalization and extra spaces
df['Sex'] = df['Sex'].str.strip().str.title()  # 'male' -> 'Male'
df['Sex']
```

### 5. Outlier Detection & Treatment (IQR Method)

* fare: Capped at the upper IQR limit so extreme ticket prices ($512) don't skew the data, without losing first-class passenger rows.
* age: Retained because elderly passengers (up to 80) are real, valid data points.

```python
# Find IQR upper limit for 'fare'
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
upper_limit = Q3 + 1.5 * (Q3 - Q1)

print('Fare outliers found:', (df['Fare'] > upper_limit).sum())

# Cap outliers at the upper limit
df['Fare'] = df['Fare'].clip(upper=upper_limit)
```

### 6. Data Type Correction

```python
df['Survived'] = df['Survived'].astype('category')
df['Pclass'] = df['Pclass'].astype('category')
df['Sex'] = df['Sex'].astype('category')
df['Embarked'] = df['Embarked'].astype('category')
df['Fare'] = df['Fare'].astype('float64')
```

### 7. Before vs. After Table & Save CSV

```python
summary = pd.DataFrame({
    'Metric': ['Rows', 'Null Values', 'Duplicates', 'Dtypes Fixed'],
    'Before': [
        len(file),
        file.isnull().sum().sum(),
        file.duplicated().sum(),#
        'No',
    ],
    'After': [len(df), df.isnull().sum().sum(), df.duplicated().sum(), 'Yes'],
})

print(summary)

# Save to CSV
df.to_csv('titanic_cleaned.csv', index=False)
```
